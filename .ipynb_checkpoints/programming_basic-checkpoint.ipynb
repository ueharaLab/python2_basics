### 3. 演算の優先度
#### 1) 数学と同じ規則。足し算よりも掛け算、掛け算よりべき乗が優先 etc  
#### 2) 整数と実数では実数が優先

````python
a=2+3*2**2
print(a)
b=(2+3)*2**2
print(b)
c=((2+3)*2)**2
print(c)
d=1.0 + 2
print(d)
````

## 3. 変数の再利用性
以下の例は、変数aを演算によって更新していくプログラムである。これに演算プログラムを追記して、aの初期値10にもどせ。ただし、演算の両辺は変数aのみを使うこと。

````python
a=8+2
print(a)
a=a**2
a=(a+a)*a
print(a)
````
